# Preprocessing for the Train Set

Missing values, outliers, unconverted categorical features, irrelavent values mislead functions and hence the model which are created with the concern of generalizability. That's why, **data preprocessing** is the most important part of machine learning process.

**Null Values**

```python
df_train.isnull().sum()
```

```python
# what is the total number of null values?

df_train.isnull().sum().sum()
```

```python
print("Mean value of each column:")

# Select only numeric columns and calculate the mean for each row
numeric_df = df_train.select_dtypes(include=["float64", "int64"])

# Calculate the mean values of each row
mean_values = numeric_df.mean(axis=0)

# Print the mean values
print(mean_values)
```

```python
print("Mean value of each column:")
numeric_df_test = df_test.select_dtypes(include=["float64", "int64"])
mean_values_test = numeric_df_test.mean(axis=0)
print(mean_values_test)
```

In the examination of the dataset, the column named as feature_04 has a different pattern compared the other columns with  numerical values. So, it is better to use the mode which is the value that appears most frequently in the column instead of the mean value of that column which is  7.725701.

```python
df_train["feature_04"].mode()

df_train["feature_04"].fillna(0, inplace=True)
```

For others, missing values is filled with the mean (average) of each numeric column in the dataset.

```python
df_train.fillna(numeric_df.mean()[:], inplace= True)

df_train.head(-1)
```

However, we only dealt with the numerical values. There are also columns with categorical values which have null values to be filled. For this end, using the mode function is again the most efficient way concerning the limited number of values of categorical columns.

```python
df_train.isnull().sum().sum()
```

```python
# Getting list of categorical variables

s = (df_train.dtypes == 'object')
object_cols = list(s[s].index)

print("Categorical variables:")
print(object_cols)
```

```python
df_train[object_cols].mode()
```

```python
# categorical value assignment

# Filling missing values with the mode for each categorical column

for col in object_cols:
    mode_value = df_train[col].mode()[0]  # Get the mode
    df_train[col].fillna(mode_value, inplace=True)  # Fill NaNs with the mode

# Checking the result

print(df_train.isnull().sum())    # Should show 0 for all categorical columns
```

# **Model Tuning and Training**

```python
# import operations of required libraries

import tensorflow as tf
from tensorflow.keras.models import  Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.utils import to_categorical

import warnings
from warnings import filterwarnings

warnings.filterwarnings('ignore', category=DeprecationWarning)
warnings.filterwarnings('ignore', category=FutureWarning)
warnings.filterwarnings('ignore', category=UserWarning)
filterwarnings('ignore')
```

```python
X = df_train.drop("target", axis=1)
y = df_train[["target"]]

# validation set

from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(X, y, 
                                                    test_size=0.20, 
                                                    random_state = 99)
num_labels = len(np.unique(y_train))  # y : labels, classes
num_labels

# Converting y_train and y_val to one-hot encoded format

y_train = to_categorical(y_train, num_classes= num_labels)
y_val = to_categorical(y_val, num_classes= num_labels)
```

```python
print(X_train.shape, " ", X_val.shape, " ")
```

```python
y_train.shape, y_val.shape
```

```python
# describe the structure of the neural network

model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=128, 
                          activation='relu',
                          name="layer1",
                          input_shape = (34,)), 
     Dropout(0.2),
    Dense(64, activation="relu"),
    Dropout(0.2),
    tf.keras.layers.Dense(units= num_labels,
                          activation='softmax', # we use sofmax function since there are classes more than two
                          name="output_layer")]
                            )

# inner structure for mathematical optimization

model.compile(loss='categorical_crossentropy',
              optimizer='adam',
              metrics=[tf.keras.metrics.Precision(),
                       tf.keras.metrics.Recall(),
                       "accuracy"]
              )

model.summary()
```

```python
history = model.fit(X_train, y_train, epochs=50, validation_data=(X_val, y_val), batch_size=32)
```
